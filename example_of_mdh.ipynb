```python
import numpy as np
from sklearn.datasets import make_blobs
from mdh import MDH  

# 1. Generate Synthetic Data (e.g., 3 clusters in 10 dimensions)
X, y = make_blobs(n_samples=500, n_features=10, centers=3, random_state=42)

# 2. Initialize the Model
# n_clusters: The target number of groups
# verbose: Print optimization progress
# plot: Show the density curve of the chosen splits
model = MDH(n_clusters=3, verbose=True, plot=True)

# 3. Fit and Predict
labels = model.fit_predict(X)

# 4. Evaluate
from sklearn.metrics import adjusted_rand_score
print(f"Clustering Score (ARI): {adjusted_rand_score(y, labels):.4f}")
```
